# **FASE 1 — Recopilación de Datos y Marco Metodológico**

```python
import pandas as pd
import glob

# ── Leer y consolidar Colombia → Canadá ──────────────────────────
archivos_can = glob.glob("comtrade_COL_CAN_*.xlsx")
archivos_can.sort()  # ordena por año automáticamente

frames_can = []
for archivo in archivos_can:
    df = pd.read_excel(archivo)
    frames_can.append(df)
    print(f"✓ {archivo} — {len(df)} filas")

consolidado_can = pd.concat(frames_can, ignore_index=True)
print(f"\nTotal Colombia → Canadá: {len(consolidado_can)} filas")

# ── Leer y consolidar Colombia → Mundo ───────────────────────────
archivos_world = glob.glob("comtrade_COL_WORLD_*.xlsx")
archivos_world.sort()

frames_world = []
for archivo in archivos_world:
    df = pd.read_excel(archivo)
    frames_world.append(df)
    print(f"✓ {archivo} — {len(df)} filas")

consolidado_world = pd.concat(frames_world, ignore_index=True)
print(f"\nTotal Colombia → Mundo: {len(consolidado_world)} filas")
```

```python
# ── Guardar archivos consolidados ────────────────────────────────
consolidado_can.to_excel(
    "comtrade_COL_CAN_2020_2024.xlsx",
    index=False,
    sheet_name="COL_CAN"
)
print("✓ Guardado: comtrade_COL_CAN_2020_2024.xlsx")

consolidado_world.to_excel(
    "comtrade_COL_WORLD_2020_2024.xlsx",
    index=False,
    sheet_name="COL_WORLD"
)
print("✓ Guardado: comtrade_COL_WORLD_2020_2024.xlsx")
```

```python
print("Columnas disponibles:")
print(consolidado_can.columns.tolist())
```

```python
# Columnas que necesitamos
columnas = ['refYear', 'cmdCode', 'cmdDesc', 'partnerDesc',
            'primaryValue', 'netWgt']

# Filtrar ambos consolidados
can_limpio = consolidado_can[columnas].copy()
world_limpio = consolidado_world[columnas].copy()

# Verificar resultado
print("Colombia → Canadá:")
print(can_limpio.head(3))
print(f"Filas: {len(can_limpio)} | Columnas: {len(can_limpio.columns)}")

print("\nColombia → Mundo:")
print(world_limpio.head(3))
print(f"Filas: {len(world_limpio)} | Columnas: {len(world_limpio.columns)}")
```

```python
print("Valores vacíos en Colombia → Canadá:")
print(can_limpio.isnull().sum())

print("\nValores vacíos en Colombia → Mundo:")
print(world_limpio.isnull().sum())
```

```python
can_limpio = can_limpio[can_limpio['primaryValue'] > 0].copy()
world_limpio = world_limpio[world_limpio['primaryValue'] > 0].copy()

print(f"Colombia → Canadá tras limpieza: {len(can_limpio)} filas")
print(f"Colombia → Mundo tras limpieza: {len(world_limpio)} filas")
```

```python
from google.colab import files

can_limpio.to_excel("COL_CAN_2020_2024_limpio.xlsx", index=False)
world_limpio.to_excel("COL_WORLD_2020_2024_limpio.xlsx", index=False)

files.download("COL_CAN_2020_2024_limpio.xlsx")
files.download("COL_WORLD_2020_2024_limpio.xlsx")
```

```python
print("Exportaciones Colombia → Canadá por año (USD):")
resumen_can = (can_limpio
               .groupby('refYear')['primaryValue']
               .sum()
               .reset_index())
resumen_can.columns = ['Año', 'Valor_USD']
resumen_can['Valor_USD_MM'] = (resumen_can['Valor_USD'] / 1_000_000).round(1)
print(resumen_can[['Año', 'Valor_USD_MM']].to_string(index=False))

print("\nExportaciones Colombia → Mundo por año (USD millones):")
resumen_world = (world_limpio
                 .groupby('refYear')['primaryValue']
                 .sum()
                 .reset_index())
resumen_world.columns = ['Año', 'Valor_USD']
resumen_world['Valor_USD_MM'] = (resumen_world['Valor_USD'] / 1_000_000).round(1)
print(resumen_world[['Año', 'Valor_USD_MM']].to_string(index=False))
```

```python
resumen_final = resumen_can[['Año', 'Valor_USD_MM']].copy()
resumen_final.columns = ['Año', 'COL_CAN_MM_USD']

# Merge resumen_final with resumen_world to align by 'Año'
resumen_final = pd.merge(resumen_final, resumen_world[['Año', 'Valor_USD_MM']], on='Año', how='left')
resumen_final.rename(columns={'Valor_USD_MM': 'COL_WORLD_MM_USD'}, inplace=True)

resumen_final['Participacion_CAN_%'] = (
    resumen_final['COL_CAN_MM_USD'] /
    resumen_final['COL_WORLD_MM_USD'] * 100
).round(2)

print(resumen_final.to_string(index=False))
```

```python
from google.colab import files

resumen_final.to_excel("tabla_01_flujos_comerciales_2020_2024.xlsx", index=False)
files.download("tabla_01_flujos_comerciales_2020_2024.xlsx")
```

```python
# ── Paso 1: Totales anuales ───────────────────────────────────────

# Total exportado por Colombia a Canadá por año (denominador izquierdo)
total_can = (can_limpio
             .groupby('refYear')['primaryValue']
             .sum()
             .rename('total_CAN'))

# Total exportado por Colombia al Mundo por año (denominador derecho)
total_world = (world_limpio
               .groupby('refYear')['primaryValue']
               .sum()
               .rename('total_WORLD'))

# ── Paso 2: Valor por producto y año ─────────────────────────────

# Exportaciones de cada producto a Canadá
x_ij = (can_limpio
        .groupby(['refYear', 'cmdCode', 'cmdDesc'])['primaryValue']
        .sum()
        .reset_index()
        .rename(columns={'primaryValue': 'X_CAN'}))

# Exportaciones de cada producto al Mundo
x_iw = (world_limpio
        .groupby(['refYear', 'cmdCode'])['primaryValue']
        .sum()
        .reset_index()
        .rename(columns={'primaryValue': 'X_WORLD'}))

# ── Paso 3: Unir todo en una sola tabla ──────────────────────────
vcr_df = x_ij.merge(x_iw, on=['refYear', 'cmdCode'], how='left')
vcr_df = vcr_df.merge(total_can, on='refYear', how='left')
vcr_df = vcr_df.merge(total_world, on='refYear', how='left')

# ── Paso 4: Calcular el VCR ──────────────────────────────────────
vcr_df['VCR'] = (
    (vcr_df['X_CAN'] / vcr_df['total_CAN']) /
    (vcr_df['X_WORLD'] / vcr_df['total_WORLD'])
).round(4)

print(f"Tabla VCR calculada: {len(vcr_df)} filas")
print(vcr_df[['refYear','cmdCode','cmdDesc','VCR']].head(10))
```

```python
# Promedio VCR 2020–2024 por producto
vcr_promedio = (vcr_df
                .groupby(['cmdCode', 'cmdDesc'])
                .agg(
                    VCR_promedio=('VCR', 'mean'),
                    VCR_max=('VCR', 'max'),
                    años_con_ventaja=('VCR', lambda x: (x > 1).sum())
                )
                .round(3)
                .reset_index()
                .sort_values('VCR_promedio', ascending=False))

# Filtrar solo productos con VCR promedio > 1
vcr_ventaja = vcr_promedio[vcr_promedio['VCR_promedio'] > 1].copy()

print(f"Productos con VCR > 1: {len(vcr_ventaja)}")
print(vcr_ventaja.head(15).to_string(index=False))
```

```python
# Filtro 1: mínimo de valor exportado a Canadá (USD 500.000 promedio anual)
valor_minimo = (can_limpio
                .groupby('cmdCode')['primaryValue']
                .mean()
                .reset_index()
                .rename(columns={'primaryValue': 'valor_promedio_anual'}))

vcr_filtrado = vcr_ventaja.merge(valor_minimo, on='cmdCode', how='left')

# Aplicar los dos filtros
vcr_filtrado = vcr_filtrado[
    (vcr_filtrado['valor_promedio_anual'] >= 500_000) &  # relevancia económica
    (vcr_filtrado['años_con_ventaja'] >= 3)              # ventaja sostenida
].copy()

vcr_filtrado = vcr_filtrado.sort_values('VCR_promedio', ascending=False)

print(f"Productos tras filtros: {len(vcr_filtrado)}")
print()
print(vcr_filtrado[['cmdCode','cmdDesc','VCR_promedio',
                     'años_con_ventaja','valor_promedio_anual']].to_string(index=False))
```

```python
from google.colab import files

# Formatear valor promedio anual en millones para legibilidad
vcr_filtrado['valor_promedio_anual_USD_MM'] = (
    vcr_filtrado['valor_promedio_anual'] / 1_000_000
).round(3)

# Tabla final limpia
vcr_final = vcr_filtrado[[
    'cmdCode',
    'cmdDesc',
    'VCR_promedio',
    'VCR_max',
    'años_con_ventaja',
    'valor_promedio_anual_USD_MM'
]].copy()

vcr_final.to_excel("vcr_productos_filtrados.xlsx", index=False)
files.download("vcr_productos_filtrados.xlsx")
print(f"✓ Archivo descargado: {len(vcr_final)} productos")
```

```python
# Scoring combinado para selección Top 10
vcr_final['score'] = (
    vcr_final['VCR_promedio'] * 0.5 +          # 50% peso al VCR
    vcr_final['años_con_ventaja'] * 2 +         # consistencia temporal
    vcr_final['valor_promedio_anual_USD_MM'] * 0.3  # relevancia económica
)

top10 = (vcr_final
         .sort_values('score', ascending=False)
         .head(10)
         .reset_index(drop=True))

top10.index += 1  # ranking desde 1

print("TOP 10 — Productos colombianos con mayor VCR hacia Canadá (2020–2024)")
print("="*70)
print(top10[['cmdCode','cmdDesc','VCR_promedio',
             'años_con_ventaja','valor_promedio_anual_USD_MM']].to_string())
```

```python
# Ver productos 11 al 15 para elegir reemplazo
reemplazo = (vcr_final
             .sort_values('score', ascending=False)
             .iloc[10:16]
             .reset_index(drop=True))

reemplazo.index += 11
print("Candidatos a reemplazar posición 10:")
print(reemplazo[['cmdCode','cmdDesc','VCR_promedio',
                  'años_con_ventaja',
                  'valor_promedio_anual_USD_MM']].to_string())
```

```python
from google.colab import files

# Códigos del Top 10 definitivo (excluye 841122, incluye 060311)
top10_codigos = [
    '90111',   # Café
    '270112',  # Carbón bituminoso
    '80450',   # Guayabas y mangos
    '60319',   # Flores frescas otras
    '300432',  # Medicamentos corticosteroides
    '611596',  # Medias sintéticas
    '730792',  # Accesorios tubería
    '30482',   # Filetes trucha
    '200819',  # Nueces y semillas
    '60311'    # Rosas frescas
]

# Construir tabla final desde vcr_final
top10_definitivo = (vcr_final[vcr_final['cmdCode']
                    .astype(str)
                    .str.strip()
                    .isin([str(c) for c in top10_codigos])]
                    .copy())

# Añadir ranking manual en el orden correcto
orden = {str(c): i+1 for i, c in enumerate(top10_codigos)}
top10_definitivo['Ranking'] = (top10_definitivo['cmdCode']
                                .astype(str)
                                .str.strip()
                                .map(orden))

top10_definitivo = (top10_definitivo
                    .sort_values('Ranking')
                    .reset_index(drop=True))

# Columnas finales limpias
top10_definitivo = top10_definitivo[[
    'Ranking', 'cmdCode', 'cmdDesc',
    'VCR_promedio', 'VCR_max',
    'años_con_ventaja', 'valor_promedio_anual_USD_MM'
]]

print("TOP 10 DEFINITIVO — MIC-CC")
print("="*60)
print(top10_definitivo.to_string(index=False))

# Guardar
top10_definitivo.to_excel("top10_definitivo_MIC_CC.xlsx", index=False)
files.download("top10_definitivo_MIC_CC.xlsx")
print("\n✓ Archivo guardado y descargado")
```
